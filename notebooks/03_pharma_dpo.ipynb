# 03 - DPO on a real pharmaceutical preference dataset

This notebook replaces the tiny synthetic preference pairs in Notebook 02 with a small public pharmaceutical-domain DPO dataset. The dataset is `ThakrePranjal/pharma-preference-dataset`, which contains 48 `prompt/chosen/rejected` preference examples for pharmaceutical QA.

The goal is still educational: learn how a real domain preference dataset changes the DPO experiment. This is not clinical validation and should not be used for medical decision-making.

## Why this experiment?

Notebook 02 used eight hand-written biomedical preference pairs. Here we use an existing pharmaceutical preference dataset and keep a held-out split.

Pipeline:

`public pharma preference data -> conversational format -> Qwen + LoRA -> DPO -> held-out evaluation`

## 1. Load the public pharmaceutical preference dataset

The source dataset has the standard DPO fields `prompt`, `chosen`, and `rejected`. It is small (48 rows), so this is intentionally a lightweight experiment.

```python
from datasets import load_dataset

RAW_DATASET = 'ThakrePranjal/pharma-preference-dataset'
raw = load_dataset(RAW_DATASET, split='train')
print(raw)
print(raw.column_names)
print(raw[0])
```

## 2. Convert to TRL conversational preference format

Qwen is an instruct/chat model. We therefore convert each textual prompt into one user message and each answer into one assistant message. TRL can then apply Qwen's chat template consistently during DPO preprocessing.

```python
from datasets import Dataset

def normalize_prompt(text):
    # The public dataset uses an Alpaca-like prompt. Keep the instruction text
    # but remove a trailing response marker when present.
    text = str(text)
    if '### Instruction:' in text:
        text = text.split('### Instruction:', 1)[1]
    if '### Response:' in text:
        text = text.split('### Response:', 1)[0]
    return text.strip()

rows = []
for row in raw:
    rows.append({
        'prompt': [{'role': 'user', 'content': normalize_prompt(row['prompt'])}],
        'chosen': [{'role': 'assistant', 'content': str(row['chosen']).strip()}],
        'rejected': [{'role': 'assistant', 'content': str(row['rejected']).strip()}],
        'topic': row.get('topic', None),
    })

dataset = Dataset.from_list(rows)
split = dataset.train_test_split(test_size=0.2, seed=42)
train_dataset = split['train']
eval_dataset = split['test']

print('Total:', len(dataset))
print('Train:', len(train_dataset))
print('Eval:', len(eval_dataset))
print('Example:', train_dataset[0])
```

## 3. Load Qwen2.5-0.5B-Instruct

We keep the same base model as Notebooks 01 and 02 so the main variable is the preference data.

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

MODEL_NAME = 'Qwen/Qwen2.5-0.5B-Instruct'
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = 'left'

dtype = torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16
model = AutoModelForCausalLM.from_pretrained(MODEL_NAME, torch_dtype=dtype)
model = model.cuda()

print('Has chat template:', tokenizer.chat_template is not None)
```

## 4. Check a held-out prompt before DPO

The baseline generation is only a qualitative reference. Later, compare this with the same prompt after DPO.

```python
def generate(model, question, max_new_tokens=120):
    messages = [{'role': 'user', 'content': question}]
    encoded = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors='pt',
        return_dict=True,
    )
    encoded = {k: v.to(model.device) for k, v in encoded.items()}
    with torch.no_grad():
        output = model.generate(
            **encoded,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            pad_token_id=tokenizer.pad_token_id,
        )
    return tokenizer.decode(output[0], skip_special_tokens=True)

TEST_INDEX = 0
test_prompt = eval_dataset[TEST_INDEX]['prompt'][0]['content']
print('QUESTION:', test_prompt)
print('
BEFORE DPO:
')
print(generate(model, test_prompt))
```

## 5. Run DPO with LoRA

Only the LoRA adapter is trained. `max_prompt_length` is intentionally omitted because the installed TRL 0.29.x API does not accept it in `DPOConfig`.

```python
from peft import LoraConfig
from trl import DPOConfig, DPOTrainer

lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    target_modules=['q_proj', 'k_proj', 'v_proj', 'o_proj'],
    bias='none',
    task_type='CAUSAL_LM',
)

dpo_args = DPOConfig(
    output_dir='./outputs/pharma-dpo',
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4,
    num_train_epochs=5,
    learning_rate=1e-5,
    beta=0.1,
    max_length=512,
    logging_steps=2,
    save_strategy='no',
    report_to='none',
    fp16=torch.cuda.is_available() and not torch.cuda.is_bf16_supported(),
    bf16=torch.cuda.is_available() and torch.cuda.is_bf16_supported(),
)

trainer = DPOTrainer(
    model=model,
    args=dpo_args,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
    processing_class=tokenizer,
    peft_config=lora_config,
)
trainer.train()
```

## 6. Evaluate preference behavior

TRL's DPO evaluation reports metrics such as chosen/rejected rewards and preference accuracy. These are more informative than looking at a single generated answer.

```python
metrics = trainer.evaluate()
print('Evaluation metrics:')
for key, value in metrics.items():
    if isinstance(value, (int, float)):
        print(f'{key}: {value:.4f}')
    else:
        print(f'{key}: {value}')
```

```python
model.eval()
print('
AFTER DPO:
')
print(generate(model, test_prompt))
```

## 7. Inspect several held-out prompts

Read the outputs critically. A lower-quality response can still look fluent, and DPO does not guarantee factual correctness.

```python
for i in range(min(5, len(eval_dataset))):
    question = eval_dataset[i]['prompt'][0]['content']
    print('=' * 80)
    print('QUESTION:', question)
    print(generate(model, question, max_new_tokens=100))
```

## 8. Save the adapter

Only the LoRA adapter is saved.

```python
ADAPTER_DIR = './outputs/pharma-dpo-adapter'
model.save_pretrained(ADAPTER_DIR)
tokenizer.save_pretrained(ADAPTER_DIR)
print('Saved:', ADAPTER_DIR)
```

### Next research direction: knowledge-grounded preferences

A more scientifically meaningful experiment would construct preference pairs from a licensed biomedical knowledge source. For example, a knowledge base can provide a verified drug-target relation as the preferred answer while a deliberately incorrect or unsupported relation is used as the rejected answer. This turns DPO into **evidence-grounded preference learning** rather than generic answer-style alignment.
